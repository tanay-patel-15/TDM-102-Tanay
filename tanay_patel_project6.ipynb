## Question 1

```python
import pandas as pd
myDF = pd.read_csv("/anvil/projects/tdm/data/flights/subset/2006.csv")
```

```python
pd.set_option("display.max_columns", None)
myDF.head()
```

```python
pa_flights = myDF[myDF["Dest"] == "PHL"]
```

```python
pa_flights["Origin"].nunique()
```

```python
phl_avg_airtime_by_origin = (
    pa_flights.groupby("Origin")["AirTime"]
    .mean()
    .sort_values(ascending=False)
)

phl_avg_airtime_by_origin.head()
```

```python
phl_avg_airtime_by_origin.loc["SFO"]
```

```python
az_flights = myDF[myDF["Dest"] == "PHX"]
```

```python
phx_avg_airtime_by_origin = (
    az_flights.groupby("Origin")["AirTime"]
    .mean()
    .sort_values(ascending=False)
)

phx_avg_airtime_by_origin.head()
```

```python
phx_avg_airtime_by_origin["HNL"]
```

In question 1, we read in the 2006 flights subset dataset using the pandas library and saved it as myDF. Then we created a subset called pa_flights that only included flights whose destination was PHL. After that, we used .nunique() to find how many unique Origin airports had flights going to PHL and used .groupby() to calculate the average AirTime for each origin airport, and we repeated a similar process for flights whose destination was PHX.

## Question 2

```python
top_20 = myDF["Dest"].value_counts().head(20).index.tolist()
top_20
```

```python
top_dests_df = myDF[myDF["Dest"].isin(top_20)]
```

```python
result = top_dests_df.groupby(["Origin", "Dest"])["AirTime"].mean()
```

```python
result.head()
```

```python
result.unstack().head()
```

In question 2, we found the 20 most popular destination airports using the .value_counts() function and saved the airport names as a list called top_20. Then we filtered the original dataframe so that it only contained flights whose destination was in the top_20 list. After that, we used .groupby() on both the Origin and Dest columns to calculate the average AirTime for each origin–destination pair and viewed the results.

## Question 4

```python
my_flights = pd.read_csv("/anvil/projects/tdm/data/flights/subset/1997.csv")
```

```python
bins = [0, 300, 600, 900, 1200, 1500, 1800, 2100, 2400]
labels = [
    "Late Night", "Early Morning", "Morning", "Late Morning",
    "Afternoon", "Early Evening", "Evening", "Night"
]
```

```python
my_flights["depart_times"] = pd.cut(my_flights["DepTime"], bins=bins, labels=labels)
```

```python
my_flights["depart_times"].value_counts(sort=False)
```

```python
my_flights["arrival_times"] = pd.cut(my_flights["ArrTime"], bins=bins, labels=labels)
```

```python
my_flights["arrival_times"].value_counts(sort=False)
```

```python
# BUSIEST DEPARTURE
my_flights["depart_times"].value_counts().idxmax()
```

```python
# BUSIEST ARRIVAL
my_flights["arrival_times"].value_counts().idxmax()
```

In question 4, we read in the 1997 flights dataset and used the .cut() function to divide the DepTime column into several time ranges based on different periods of the day. We assigned labels such as Late Night, Early Morning, and Afternoon to these bins and saved the results as the new column depart_times. Then we repeated the same process for the ArrTime column to create the arrival_times column and examined which time ranges had the most flights.
