In Pandas, we usually filter and select columns using bracket-based indexing like df[df["col"] == value] or df[["col1", "col2"]]. In Polars, we explicitly use expressions like .filter(pl.col("col") == value) and .select(pl.col("col")), which makes the operations more structured and clear.

```python
flights_2005_path = "/anvil/projects/tdm/data/flights/subset/2005.csv"
myDF = pl.read_csv(flights_2005_path, null_values=["NA"])
```

```python
indy_flights = myDF.filter(pl.col("Origin") == "IND")
```

```python
indy_flights_crs = indy_flights.select(pl.col("^CRS.*$"))
indy_flights_crs.head()
```

In question 3, we read in the 2005 Flights dataset using Polars and filtered the data to only include flights that originated from IND. Then we used the .select() function with a pl.col() expression to display only the columns whose names begin with CRS. This allowed us to focus on the scheduled time-related columns for the selected flights.

```python
myDF = myDF.with_columns(
    pl.col("ArrTime").cast(pl.Int64, strict=False)
)
```

```python
myDF = myDF.with_columns(
    (pl.col("ArrTime") - pl.col("CRSArrTime")).abs().alias("AbsDiffArrTime")
)
```
