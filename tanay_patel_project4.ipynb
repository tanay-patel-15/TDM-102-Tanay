## Question 5

```python
children = myDF["Age"][myDF["Age"] <= 20].tolist()
young    = myDF["Age"][(myDF["Age"] > 20) & (myDF["Age"] <= 40)].tolist()
adult    = myDF["Age"][(myDF["Age"] > 40) & (myDF["Age"] <= 60)].tolist()
old      = myDF["Age"][myDF["Age"] > 60].tolist()
```

```python
myDF["AgeGroup"] = None
myDF.loc[myDF["Age"].isin(children), "AgeGroup"] = "Children"
myDF.loc[myDF["Age"].isin(young),    "AgeGroup"] = "Young"
myDF.loc[myDF["Age"].isin(adult),    "AgeGroup"] = "Adult"
myDF.loc[myDF["Age"].isin(old),      "AgeGroup"] = "Old"
```

```python
myDF["AgeGroup"].value_counts()
```

```python
pd.crosstab(myDF["AgeGroup"], myDF["Combined"])
```

In question 5, we used the people list and created three more lists based on age ranges: children, young, and adult, and we also used old for ages above 60. Then we created a new column called AgeGroup and used .isin() with .loc[] to assign a label to each person’s Age based on which list their age belongs to. Finally, we displayed the value counts of AgeGroup and used pd.crosstab() to compare the combinations between AgeGroup and the Combined column.
